```python
def imprimir_bitboard(bitboard):
    """Imprime el bitboard en una cuadrícula de 8x8 usando np.binary_repr()."""
    bits = np.binary_repr(bitboard, width=64)  # Invertimos el orden de los bits
    for fila in range(8):  # Recorremos de arriba hacia abajo
        print(" ".join(bits[(fila) * 8:(fila + 1) * 8]), flush=True)  # Dividimos en filas de 8
```

```python
def mask_bishop_attacks(square):
    # Inicializa o bitboard de ataques
    attacks = 0

    # Calcula a linha (rank) e coluna (file) do quadrado
    tr = square // 8
    tf = square % 8

    # Máscara para as diagonais superiores (direita e esquerda)
    r, f = tr + 1, tf + 1
    while r <= 6 and f <= 6:
        attacks |= (1 << (r * 8 + f))
        r += 1
        f += 1

    r, f = tr + 1, tf - 1
    while r <= 6 and f >= 1:
        attacks |= (1 << (r * 8 + f))
        r += 1
        f -= 1

    # Máscara para as diagonais inferiores (direita e esquerda)
    r, f = tr - 1, tf + 1
    while r >= 1 and f <= 6:
        attacks |= (1 << (r * 8 + f))
        r -= 1
        f += 1

    r, f = tr - 1, tf - 1
    while r >= 1 and f >= 1:
        attacks |= (1 << (r * 8 + f))
        r -= 1
        f -= 1

    # Retorna o mapa de ataques
    return attacks

imprimir_bitboard(mask_bishop_attacks(2))
print("\n")
imprimir_bitboard(mask_bishop_attacks(15))
print("\n")
imprimir_bitboard(mask_bishop_attacks(43))
print("\n")
imprimir_bitboard(mask_bishop_attacks(59))
```

```python
def mask_rook_attacks(square):
    # Inicializa o bitboard de ataques
    attacks = 0

    # Calcula a linha (rank) e coluna (file) do quadrado
    tr = square // 8
    tf = square % 8

    # Máscara para as linhas e colunas
    for r in range(tr + 1, 7):
        attacks |= (1 << (r * 8 + tf))

    for r in range(tr - 1, 0, -1):
        attacks |= (1 << (r * 8 + tf))

    for f in range(tf + 1, 7):
        attacks |= (1 << (tr * 8 + f))

    for f in range(tf - 1, 0, -1):
        attacks |= (1 << (tr * 8 + f))

    # Retorna o mapa de ataques
    return attacks

imprimir_bitboard(mask_rook_attacks(2))
print("\n")
imprimir_bitboard(mask_rook_attacks(15))
print("\n")
imprimir_bitboard(mask_rook_attacks(43))
print("\n")
imprimir_bitboard(mask_rook_attacks(59))
```

```python
# pawns_bitboard = 65280
# rock_bitboard = 129

bitboards = {"pawns" : 65280, "rock" : 129}

bitboard = bitboards["pawns"] | bitboards["rock"]

print(np.binary_repr( bitboard, width=64))

# Movimiento de una casilla a otra
origen = 0
destino = 1

origen_mask = 1 << origen
newbitboard = bitboard ^ origen_mask # Quitamos la ficha del origen
shift = 1 << destino # Movemos la ficha al destino
newbitboard = newbitboard | shift # Ponemos la ficha en el destino

print(np.binary_repr( newbitboard, width=64))

for i in bitboards:
    if bitboards[i] & origen_mask:
        print("Ficha encontrada {}".format(i))
        break
```
