```python
def imprimir_bitboard(bitboard):
    """Imprime el bitboard en una cuadrícula de 8x8 usando np.binary_repr()."""
    bits = np.binary_repr(bitboard, width=64)  # Invertimos el orden de los bits
    for fila in range(8):  # Recorremos de arriba hacia abajo
        print(" ".join(bits[(fila) * 8:(fila + 1) * 8]), flush=True)  # Dividimos en filas de 8
```

```python
# pawns_bitboard = 65280
# rock_bitboard = 129

bitboards = {"pawns" : 65280, "rock" : 129}

bitboard = bitboards["pawns"] | bitboards["rock"]

print(np.binary_repr( bitboard, width=64))

# Movimiento de una casilla a otra
origen = 0
destino = 1

origen_mask = 1 << origen
newbitboard = bitboard ^ origen_mask # Quitamos la ficha del origen
shift = 1 << destino # Movemos la ficha al destino
newbitboard = newbitboard | shift # Ponemos la ficha en el destino

print(np.binary_repr( newbitboard, width=64))

for i in bitboards:
    if bitboards[i] & origen_mask:
        print("Ficha encontrada {}".format(i))
        break
```
